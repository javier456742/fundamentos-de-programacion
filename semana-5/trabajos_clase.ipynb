```python
desordenada = (9, 2, 7, 10, 0, 8, 14, 22, 4)

print(type(desordenada))
print(desordenada)

lista = list(desordenada)

print()
print(type(lista))
print(lista)

print()
lista.sort()
print(lista)

ordenada = tuple(lista)

print()
print(type(ordenada))
print(ordenada)
```

```python
def suma_tupla(t):
    total = 0
    for e in t:
        total += e
    return total

calorias = (800, 950, 700, 1300, 800, 900, 650)

resultado = suma_tupla(calorias)

print(f"Metiste a tu cuerpo {resultado} calorias, gordo")
```

```python
contacto = {
    "Ricardo": "4435734298",
    "Pedro": "3335734298",
    "Luisa": "6635734298",
}

contacto["Cris"] = "446794728"
contacto["Ricardo"] = "449794728"

print(contacto)
```

```python
print(contacto.keys())
print(contacto.values())
print(contacto.items())
```

```python
print("Impresion de llaves")
for n in contacto.keys():
    print(n)

print("\nImpresion de valores")
for v in contacto.values():
    print(v)

print("\nImpresion de completa")
for n,t  in contacto.items():
    print(f"Nombres: {n}, Telefono: {t}")

```

```python
def buscar_contacto(contacto, nombre):
    return contacto.get(nombre, "contacto no encontrado")

nombre = input("Escribe el nombre del contacto que buscas")
telefono = buscar_contacto(contacto, nombre)

print(f"El telefono de {nombre} es: {telefono}")
```
