### Desafío Extra 1: Lista de espera con tuplas
Crea una tupla `espera = ("María", "José", "Carlos", "Lucía", "Pedro")`. Imprime el tercer elemento, captura dos nombres nuevos con `input()`, anéxalos en una nueva tupla, conviértela en lista, ordénala alfabéticamente y muestra la cantidad total de personas.


```python
espera = ("María", "José", "Carlos", "Lucía", "Pedro")

print(f"El tercer elemento de la lista de espera es: {espera[2]}")

nombre1 = input("Ingresa el primer nombre nuevo: ")
nombre2 = input("Ingresa el segundo nombre nuevo: ")

nueva_tupla = espera + (nombre1, nombre2)

lista_espera = list(nueva_tupla)
lista_espera.sort()

print("\nLista de espera ordenada alfabéticamente:")
print(lista_espera)
print(f"Cantidad total de personas: {len(lista_espera)}")
```

### Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con `items()` el color ganador junto con su total.

```python
candidatos = {
    "rojo" : 0,
    "azul" : 0,
    "verde" : 0
}

cantidad = int(input("¿Cuántos votos van a ser?"))
rest = 0

while cantidad != rest:
    voto = input(f"Ingresa el voto {rest+1} (rojo, verde y azul): ")
    print(f"Votaste por {voto.lower()}")
    if voto in candidatos:
        print("El voto está dentro de los candidatos")
        candidatos[voto.lower()] += 1
        rest += 1
    else:
        print("El voto no es válido")
    
        

for k, v in candidatos.items():
    print(f"Opción {k}, votos: {v}")
```

### Desafío Extra 3: Validación robusta de entrada
Escribe un programa que pida al usuario dos números enteros dentro de un bloque `try-except`. Si el usuario ingresa texto o un valor vacío, muestra un mensaje controlado y repite la solicitud con un ciclo `while`. Luego pide la división del primero entre el segundo, con un bloque específico para `ZeroDivisionError`.

```python
def pedir_entero(mensaje):
    while True:
        try:
            valor = int(input(mensaje))
            return valor
        except ValueError:
            print("Error: Debes ingresar un número entero válido (no texto ni campos vacíos).\n")

dividendo = pedir_entero("Ingresa el dividendo entero: ")
divisor = pedir_entero("Ingresa el divisor entero: ")

try:
    resultado = dividendo / divisor
    print(f"El resultado de la división {dividendo} / {divisor} es: {resultado}")
except ZeroDivisionError:
    print("Error: No es posible dividir entre cero.")
```

### Desafío Extra 4: Analizador de textos
Crea un programa que reciba una frase, imprima su longitud con `len()`, la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo.

```python
def contar_palabras(frase):
    palabras = frase.split()
    return len(palabras)

frase_usuario = input("Ingresa una frase: ")

print(f"Longitud total del texto: {len(frase_usuario)} caracteres")

print(f"Texto en mayúsculas: {frase_usuario.upper()}")

frase_modificada = frase_usuario.replace("Python", "programación").replace("python", "programación")
print(f"Texto reemplazado: {frase_modificada}")

total_palabras = contar_palabras(frase_usuario)
print(f"Cantidad total de palabras: {total_palabras}")
```

### Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo `while` y opciones numéricas que permita ejecutar de forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en el diccionario `contactos`, (3) dividir dos números con manejo de `ZeroDivisionError` y (4) analizar un mensaje de texto. La opción (5) debe finalizar el programa.

```python
def opcion_1():
    numeros = (15, 3, 8, 42, 1, 23)
    lista_ordenada = sorted(numeros)
    print("\n--- Opción 1: Tupla ordenada ---")
    print(f"Tupla original: {numeros}")
    print(f"Lista ordenada: {lista_ordenada}")

def opcion_2():
    contactos = {"Juan": "555-1234", "María": "555-5678", "Pedro": "555-9012"}
    nombre = input("\nIngresa el nombre del contacto a buscar: ").strip().capitalize()
    print("--- Opción 2: Búsqueda de contactos ---")
    if nombre in contactos:
        print(f"El teléfono de {nombre} es: {contactos[nombre]}")
    else:
        print(f"No se encontró el contacto '{nombre}'.")

def opcion_3():
    print("\n--- Opción 3: División de números ---")
    try:
        dividendo = float(input("Ingresa el dividendo: "))
        divisor = float(input("Ingresa el divisor: "))
        resultado = dividendo / divisor
        print(f"El resultado es: {resultado}")
    except ZeroDivisionError:
        print("No se puede dividir entre cero.")
    except ValueError:
        print("Entrada inválida. Debes ingresar números.")

def opcion_4():
    print("\n--- Opción 4: Análisis de texto ---")
    texto = input("Ingresa una frase para analizar: ")
    cant_palabras = len(texto.split())
    print(f"El mensaje contiene {len(texto)} caracteres y {cant_palabras} palabras.")

opcion = ""
while opcion != "5":
    print("\n" + "="*30)
    print("             MENU")
    print("="*30)
    print("1. Mostrar tupla de números ordenada")
    print("2. Buscar teléfono en contactos")
    print("3. Dividir dos números")
    print("4. Analizar mensaje de texto")
    print("5. Salir")
    
    opcion = input("Selecciona una opción (1-5): ").strip()
    
    if opcion == "1":
        opcion_1()
    elif opcion == "2":
        opcion_2()
    elif opcion == "3":
        opcion_3()
    elif opcion == "4":
        opcion_4()
    elif opcion == "5":
        print("\nGracias por usar el programa")
    else:
        print("\nOpción no válida")
```
