# Ejercicio 10: Algoritmo para entrada a una película
**Enunciado:** Diseña un algoritmo que reciba la edad de una persona y decida si
puede entrar a una película para mayores de 18 años.

```python
edad = int(input("Ingresa tu edad(debes ser mayor de 18 para entrar): "))

if edad >= 18:
    print("Puedes entrar a la funcion")
else:
    print("No cumple con la edad requerida")
```

### Ejercicio 7: Algoritmo para saber si un número es par
**Enunciado:** Escribe los pasos del algoritmo (en lenguaje natural) que reciba un
número entero y determine si es par o impar. La paridad se verifica con la
operación `numero % 2`

```python
numero = int(input("Ingresa un numero entero: "))

nf = numero%2

if nf == 0:
    print(f"El numero {numero} es par")
else:
    print(f"El numero {numero} es impar")
```

### Ejercicio 16: Intercambio de variables en PSeInt
**Enunciado:** Escribe el pseudocódigo que lea dos valores y los intercambie usando
una variable temporal.


```python
a = 3
b = 8

print(f"a={a}, b={b}")

c = a
a = b
b = c

print(f"a={a}, b={b}")
```
