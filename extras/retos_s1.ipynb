# Algoritmo para calcular velocidad

```python
distancia = int(input("Ingrese la distancia en kilómetros: "))
tiempo = int(input("Ingrese el tiempo en horas: "))

print("El dato distancia es de tipo: ", type(distancia))
print("El dato tiempo es de tipo: ", type(tiempo))

velocidad = distancia / tiempo

print("La velocidad promedio es:", velocidad)
```

# Promedio

```python
c1 = float(input("Dame la primera calificacion"))
c2 = float(input("Dame la sugunda calificacion"))
c3 = float(input("Dame la tercera calificacion"))

promedio = ((c1+c2+c3)/3)

print(f"Tu promedio es: {promedio}")
```

# Grados Fahrenheit 

```python
print("convertidor de celsius a Fahrenheit ")

cel = float(input("inserta los grados celsius que quieras aa Fahrenheit: "))

Fahrenheit = (cel*(9/5)) + 32

print(f"{cel} grados equivalen a {Fahrenheit} grados Fahrenheit")
```
