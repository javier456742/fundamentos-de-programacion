### Ejercicio 1: Mayor de edad


```python
edad = int(input("Ingresa tu edad:"))

if(edad>=18):
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
```

### Ejercicio 2: Número par o impar


```python
numero = int(input("Ingresa un numero entero"))

if(numero%2 == 0):
    print("Es numero par")
else:
    print("Es un numero impar")
```

### Ejercicio 3: Calificación aprobatoria

### Ejercicio 4: Clasificación por rango de edad con elif

```python
edad = int(input("Ingresa tu edad:"))

if(edad >= 18):
    print("Eres adulto")
elif(edad >= 13):
    print("Eres adolescente")
elif(edad >= 3):
    print("Eres ninio")
else:
    print("Bebe")
```

### Ejercicio 5: Descuento por tipo de visitante


```python
visitante = input("Que tipo de visitante es?\n1.-Adulto mayor\n2.-Profesor\n3.-Estudiante\n4.-Otro")
boleto = 45

if(visitante == "Adulto mayor"):
    pf=boleto-(boleto*.12)
    print(f"Precio final: {pf}")
elif(visitante == "Profesor"):
    pf=boleto-(boleto*.10)
    print(f"Precio final: {pf}")
elif(visitante == "Estudiante"):
    pf=boleto-(boleto*.10)
    print(f"Precio final: {pf}")
else:
    print(f"Precio final: {boleto}")
```

### Ejercicio 7: Detección de número negativo con or

```python
n1 = int(input("Dame un numero: "))
n2 = int(input("Dame un segundo numero: "))

if(n1<0 or n2<0):
    print("Hay un numero negativo")
else:
    print("Los dos son positivos")
```

### Ejercicio 8: Positivo y par con anidamiento

```python
n = int(input("Dame un numero: "))

if(n>0):
    print("Es positivo")
    if(n%2 == 0):
        print("Es par")
    else:
        print("Es impar")
else:
    print("Es negativo")
```

### Ejercicio 9: Salir del proceso con not

```python
cont = input("Deseas continuar")

if(not (cont == "no")):
    print("Proceso finalizado")
else:
    n = int(input("Dame un numero: "))
    if(n>0):
        print("Es positivo")
        if(n%2 == 0):
            print("Es par")
        else:
            print("Es impar")
    else:
        print("Es negativo")
```

### Ejercicio 11: Suma de los primeros N números

```python
limit = int(input("Dame un numero:"))

sumat = 0
cont = 0

while cont <= limit:
    sumat += cont
    cont += 1

print(sumat)
```
