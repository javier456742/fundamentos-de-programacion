# Funciones que no regresan valor

```python
def menu():
    print("Esto va a ser un menu")
    print("1. Opcion 1")
    print("2. Opcion 2")
    print("3. Opcion 3")
    
print(menu())
```

```python
def suma(a,b):
    c = a + b
    return c

print(suma(5,10))
```

```python
def promedio(sumatoria, n):
    return sumatoria/n

n = int(input("¿Cuantas calificaciones quiere promediar?: "))

sumatoria = 0
print(f"Se van a pedir {n} calificaciones")
for i in range(1, n+1):
    calif = float(input(f"Introduce la calificación {i}: "))
    sumatoria = sumatoria + calif

print(f"El promedio de las {n} calificaciones es: {promedio(sumatoria, n)}")
```
