```python
def promedio(sumatoria, n):
    return sumatoria/n

n = int(input("¿Cuantas calificaciones quiere promediar?: "))

sumatoria = 0
print(f"Se van a pedir {n} calificaciones")
for i in range(1, n+1):
    calif = float(input(f"Introduce la calificación {i}: "))
    sumatoria = sumatoria + calif

print(f"El promedio de las {n} calificaciones es: {promedio(sumatoria, n)}")
```

```python
filas = int(input("Introduce el número de filas: "))
columnas = int(input("Introduce el número de columnas: "))

matriz = []

for f in range(filas):
    fila_actual = []
    for c in range(columnas):
        valor = int(input(f"Introduce el valor para la posición [{f}][{c}]: "))
        fila_actual.append(valor)
    matriz.append(fila_actual)
print(matriz)

print("matriz completa:")
for fila in matriz:
    for columna in fila:
        print(columna, end=" ")
    print()

```

```python
n = int(input("Introduce el número : "))

matriz = []
i=1

for f in range(n):
    fila_actual = []
    for c in range(n):
        inv = n - i
        if(c == inv):
            fila_actual.append(1)
            i += 1
        elif(c==f):
            fila_actual.append(1)
        else:
            fila_actual.append(0)

    matriz.append(fila_actual)
print(matriz)

print("matriz completa:")
for fila in matriz:
    for columna in fila:
        print(columna, end=" ")
    print()
```
