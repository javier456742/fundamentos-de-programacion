Ejercicio 6: División con restas sucesivas
Enunciado: Sin usar los operadores / ni //, calcula el cociente y el residuo de una división entre enteros positivos restando el divisor al dividendo con un while; el cociente es la cantidad de restas realizadas. Entrada: 17 y 5 Proceso: 17-5=12, 12-5=7, 7-5=2 (3 restas, ya no se puede restar más) Salida: 17 ÷ 5 = 3 con residuo 2



```python
dividendo = int(input("Dame el dividendo: "))
divisor = int(input("Ingresa uel divisor: "))

if divisor <= 0:
    print("El número debe ser mayor a 0")
else:
    cociente = 0
    residuo = dividendo

    while residuo >= divisor:
        residuo = residuo - divisor
        cociente = cociente + 1

    print(f"{dividendo} / {divisor} = {cociente} con residuo {residuo}")
    print(f"Termino el proceso")
```
