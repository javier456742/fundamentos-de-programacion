# Extra 1: División de cuenta con propina

```python
c = float(input("Cual es el total de la cuenta?"))
porc = float(input("Cuanto porcentaje de propina quiere dar?"))
personas = float(input("Entre cuantas personas se dividira la cuenta?"))

prop = (15*0.01)*c
ct = prop+c
tp = ct/personas

print(f"Propina:{prop:.2f}\n Total con propina:{ct:.2f}\n Pago por persona:{tp:.2f}")
```

# Extra 2: Conversor de minutos a días, horas y minutos

```python
mp = int(input("Cuantos minutos quieres convertir"))

minutosdia = 1440

dias = mp // minutosdia
minrest = mp % minutosdia

horas = minrest//60
minrest2 = minrest%60

print(f"{mp} minutos: {dias} dia(s), {horas} hora(s), {minrest2} minuto(s)")
```

# Extra 3: Calificación final ponderada

```python
p1 = int(input("Dame la calificacion tu primer parcial(30%): "))
p2 = int(input("Dame la calificacion tu segundo parcial(30%): "))
p3 = int(input("Dame la calificacion tu tercer parcial(30%): "))

cf = (p1*.3)+(p2*.3)+(p3*.4)

print(f"Tu calificacion final es: {cf:.2f}")
```

# Extra 4: Conversor de moneda (MXN a USD y EUR)

```python
pm = float(input("Cantidad en pesos mexicanos:"))
cd = float(input("Tipo de cambio USD:"))
ce = float(input("Tipo de cambio EUR:"))

dolar = pm/cd
euro = pm/ce

print(f"{pm:.2f} mx equivalen a: \nTipo de cambio USD: {dolar:.2f} \nTipo de cambio EUR: {euro:.2f}")

```
